```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np
```

```python
webpage=requests.get('https://books.toscrape.com/catalogue/page-2.html')
```

```python
webpage.encoding = 'utf-8'
```

```python
soup=BeautifulSoup(webpage.text,'lxml')

# print(soup.prettify())
```

```python
book_names = []
for book in soup.select('article.product_pod'):
    book_names.append(book.h3.a['title'])  # Access the title attribute directly

book_names
```

```python
prices = soup.find_all('p', class_='price_color')
```

```python
price_list = []
for price in prices:
    price_list.append(price.text)
```

```python
pd.DataFrame({'name': book_names, 'Prices':price_list})
```

```python
final_df = pd.DataFrame()
for j in range(1, 50):
    webpage = requests.get(f'https://books.toscrape.com/catalogue/page-{j}.html')
    webpage.encoding = 'utf-8'
    soup = BeautifulSoup(webpage.text, 'lxml')
    book_names = []
    for book in soup.select('article.product_pod'):
        book_names.append(book.h3.a['title'])

    prices = soup.find_all('p', class_='price_color')
    price_list = []
    for price in prices:
        price_list.append(price.text)

    image = []
    for i in soup.find_all('img'):
        image.append(i['src'])



    stock = []
    for s in soup.find_all('p', class_='instock availability'):
        stock.append(s.text.strip())



    df = pd.DataFrame({'Name': book_names, 'Prices': price_list, 'Image':image, 'Stock': stock})
    final_df = pd.concat([final_df, df], ignore_index=True)




```

```python

final_df['Stock'] = final_df['Stock'].apply(lambda x: x.replace('In stock', 'Yes'))
```

```python

final_df['Stock'] = final_df['Stock'].apply(lambda x: x.replace('False', 'No'))
```
